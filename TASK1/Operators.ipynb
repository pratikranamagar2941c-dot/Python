### Operators in Python

Operators are symbols or keywords that tell Python to perform an action on one or more values. We use them in everyday programs for calculations, comparisons, checking conditions, and working with data. For example, `+` adds numbers, `==` checks whether two values are equal, and `and` combines conditions.

Python has several groups of operators. In this assignment, I will cover arithmetic, assignment, comparison, logical, membership, identity, and bitwise operators, along with operator precedence.

#### 1. Arithmetic operators

Arithmetic operators are used to perform mathematical calculations. These are similar to the symbols we use in normal mathematics, although Python also has operators for floor division and finding a remainder.

| **Operator** | **Name** | **Example** | **Result** |
|---|---|---|---|
| `+` | Addition | `10 + 3` | `13` |
| `-` | Subtraction | `10 - 3` | `7` |
| `*` | Multiplication | `10 * 3` | `30` |
| `/` | Division | `10 / 3` | `3.333...` |
| `//` | Floor division | `10 // 3` | `3` |
| `%` | Modulus (remainder) | `10 % 3` | `1` |
| `**` | Exponentiation (power) | `2 ** 3` | `8` |

**Syntax:**
```python
result = value1 operator value2
```

**Example:**

````python
a = 10
b = 3

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Division
print(a // b)  # Floor division
print(a % b)   # Remainder
print(a ** b)  # Power
````

**Output:**
```text
13
7
30
3.3333333333333335
3
1
1000
```

A useful difference to remember is that `/` gives a division result as a float, while `//` gives the floor of the division. The `%` operator gives the remainder. For example, `10 % 3` is `1` because 3 goes into 10 three times with 1 left over.

#### 2. Assignment operators

Assignment operators are used to give a value to a variable or update the value it already has. The basic assignment operator is `=`, but Python also has shorthand operators that combine a calculation with assignment.

| **Operator** | **Example** | **Same as** |
|---|---|---|
| `=` | `x = 5` | Assign 5 to `x` |
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 3` | `x = x - 3` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 3` | `x = x / 3` |
| `//=` | `x //= 3` | `x = x // 3` |
| `%=` | `x %= 3` | `x = x % 3` |
| `**=` | `x **= 3` | `x = x ** 3` |
| `&=`, `|=`, `^=` | `x &= 3` | Bitwise update |
| `>>=`, `<<=` | `x <<= 1` | Bit shift update |

**Syntax:**
```python
variable = value
variable operator= value
```

**Example:**

````python
score = 10
print(score)

score += 5
print(score)

score *= 2
print(score)

score -= 4
print(score)
````

**Output:**
```text
10
15
30
26
```

The shorthand operators make it quicker to update a variable. For example, `score += 5` adds 5 to the current value of `score` and stores the new result back in the same variable.

#### 3. Comparison operators

Comparison operators compare two values. The result of a comparison is always a Boolean value: `True` or `False`. They are often used in `if` statements and loops to help a program make decisions.

| **Operator** | **Meaning** | **Example** | **Result** |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal to | `5 >= 5` | `True` |
| `<=` | Less than or equal to | `3 <= 5` | `True` |

**Syntax:**
```python
value1 comparison_operator value2
```

**Example:**

````python
a = 15
b = 10

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= 15)
print(b <= 10)
````

**Output:**
```text
False
True
True
False
True
True
```

Python also allows chained comparisons. For example, `10 < score < 20` checks whether `score` is greater than 10 and less than 20.

````python
score = 15
print(10 < score < 20)
````

**Output:**
```text
True
```

#### 4. Logical operators

Logical operators are used to combine conditions or reverse a condition. They are especially useful when a program needs to check more than one thing at the same time.

| **Operator** | **Meaning** | **Description** |
|---|---|---|
| `and` | Logical AND | True only when both conditions are true. |
| `or` | Logical OR | True when at least one condition is true. |
| `not` | Logical NOT | Reverses the Boolean result. |

**Syntax:**
```python
condition1 and condition2
condition1 or condition2
not condition
```

**Example:**

````python
age = 20
has_id = True

print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
````

**Output:**
```text
True
True
False
```

In the first example, both conditions are true, so `and` returns `True`. In the second example, at least one condition is true, so `or` returns `True`. The `not` operator changes `True` into `False` and `False` into `True`.

Python's `and` and `or` operators use short-circuit evaluation. This means Python may stop checking once the final result is already known. They can also return one of their operands, not always just `True` or `False`.

#### 5. Membership operators

Membership operators check whether a value is present in a sequence or collection, such as a string, list, tuple, or set. They return `True` or `False`.

| **Operator** | **Meaning** | **Description** |
|---|---|---|
| `in` | Present in | Checks whether a value exists in a collection. |
| `not in` | Not present in | Checks whether a value does not exist in a collection. |

**Syntax:**
```python
value in collection
value not in collection
```

**Example:**

````python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)
print("orange" in fruits)
print("orange" not in fruits)

word = "Python"
print("th" in word)
````

**Output:**
```text
True
False
True
True
```

Membership checks are case-sensitive for strings. For example, `"P" in "Python"` is `True`, but `"p" in "Python"` is `False`.

#### 6. Identity operators

Identity operators check whether two variables refer to the exact same object in memory. They are different from comparison operators, which check whether values are equal.

| **Operator** | **Meaning** | **Description** |
|---|---|---|
| `is` | Same object | Returns `True` if both references point to the same object. |
| `is not` | Different objects | Returns `True` if they point to different objects. |

**Syntax:**
```python
object1 is object2
object1 is not object2
```

**Example:**

````python
first_list = [1, 2, 3]
second_list = first_list
third_list = [1, 2, 3]

print(first_list == third_list)
print(first_list is second_list)
print(first_list is third_list)
print(first_list is not third_list)
````

**Output:**
```text
True
True
False
True
```

Here, `first_list` and `third_list` contain equal values, so `==` returns `True`. However, they are two separate list objects, so `is` returns `False`. The variable `second_list` refers to the same list object as `first_list`, so `is` returns `True`.

A common use of identity checking is checking whether a value is `None`, for example: `result is None`.

#### 7. Bitwise operators

Bitwise operators work on the binary (0 and 1) form of integers. They compare or move individual bits. These operators are used in areas such as low-level programming, working with flags, and some networking tasks.

| **Operator** | **Name** | **Description** |
|---|---|---|
| `&` | AND | Sets a bit to 1 when both corresponding bits are 1. |
| `|` | OR | Sets a bit to 1 when at least one corresponding bit is 1. |
| `^` | XOR | Sets a bit to 1 when the corresponding bits are different. |
| `~` | NOT | Inverts the bits of an integer. |
| `<<` | Left shift | Shifts bits to the left. |
| `>>` | Right shift | Shifts bits to the right. |

**Syntax:**
```python
value1 & value2
value1 | value2
value1 ^ value2
~value
value << number_of_positions
value >> number_of_positions
```

**Example:**

````python
a = 5   # Binary: 0101
b = 3   # Binary: 0011

print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 1)
print(a >> 1)
````

**Output:**
```text
1
7
6
-6
10
2
```

For the first three operations, we can look at the binary values:

| Operation | Binary calculation | Result |
|---|---|---|
| `5 & 3` | `0101 & 0011 = 0001` | `1` |
| `5 | 3` | `0101 | 0011 = 0111` | `7` |
| `5 ^ 3` | `0101 ^ 0011 = 0110` | `6` |

The `~` operator inverts an integer's bits. In Python, `~5` gives `-6`. For positive integers, left shift by one position is like multiplying by 2, and right shift by one position is like floor-dividing by 2.

#### 8. Operator precedence

Operator precedence decides the order in which Python evaluates operators in an expression. It is similar to the order of operations we use in mathematics. Operators with higher precedence are evaluated before operators with lower precedence. Parentheses can be used to make the order clear or to change it.

**Example:**

````python
result1 = 10 + 2 * 3
result2 = (10 + 2) * 3

print(result1)
print(result2)
````

**Output:**
```text
16
36
```

In the first expression, multiplication happens before addition, so the answer is `16`. In the second expression, parentheses are evaluated first, so the answer is `36`.

##### Python operator precedence (highest to lowest)

| **Order** | **Operators** | **Description** |
|---:|---|---|
| 1 | `()` | Parentheses / grouping |
| 2 | `**` | Exponentiation |
| 3 | `+x`, `-x`, `~x` | Unary plus, minus, and bitwise NOT |
| 4 | `*`, `/`, `//`, `%` | Multiplication, division, floor division, modulus |
| 5 | `+`, `-` | Addition and subtraction |
| 6 | `<<`, `>>` | Bitwise shifts |
| 7 | `&` | Bitwise AND |
| 8 | `^` | Bitwise XOR |
| 9 | `|` | Bitwise OR |
| 10 | `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `is not`, `in`, `not in` | Comparisons, identity, and membership |
| 11 | `not` | Logical NOT |
| 12 | `and` | Logical AND |
| 13 | `or` | Logical OR |

Assignment statements such as `x = 5` and augmented assignments such as `x += 2` are not ordinary expression operators in this precedence table. When an expression is difficult to read, using parentheses is a good habit.

##### More examples of operator precedence

These examples show how precedence works with different groups of operators.

````python
print(2 + 3 * 4)       # Multiplication first
print((2 + 3) * 4)     # Parentheses first
print(2 ** 3 ** 2)     # Exponentiation is evaluated from right to left
print(True or False and False)  # 'and' is evaluated before 'or'
````

**Output:**
```text
14
20
512
True
```

The third result is `512` because Python evaluates `3 ** 2` first, then calculates `2 ** 9`. The last result is `True` because `and` has higher precedence than `or`.
