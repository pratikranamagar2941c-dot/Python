# Python Functions

**Assignment notes and practical examples**

Functions are one of the most useful parts of Python. They let us group instructions under a meaningful name and reuse those instructions whenever we need them. In this notebook, I have explained the main ideas about functions with short examples and comments.

The examples use simple situations such as greetings, student details, scores, and calculations so that each concept is easier to follow.



## 1. What is a function?

A **function** is a reusable block of code that performs a particular task. Instead of writing the same instructions again and again, we can put them inside a function and call that function whenever it is needed.

A function is not executed just because it has been defined. We need to call it by writing its name followed by parentheses.

````python
def greet():
    print("Hello, welcome to Python!")

# Calling the function
greet()
greet()
````

### The basic syntax

```python
def function_name():
    # statements that belong to the function
    pass
```

- `def` is the keyword used to define a function.
- `function_name` is the name we choose for the function.
- Parentheses `()` may contain parameters.
- The function body is indented.
- `pass` is a placeholder that does nothing; it can be used while planning a function.

## 2. Parameters and arguments

These terms are connected, but they mean different things:

- A **parameter** is the variable written in the function definition.
- An **argument** is the actual value supplied when the function is called.

For example, `name` is a parameter in the function below. `"Ram"` and `"Sita"` are arguments in the two function calls.

````python
def greet(name):
    print("Hello", name)

greet("Ram")
greet("Sita")
````

### Functions with more than one parameter

A function can accept several parameters. When we pass values by position, Python assigns them from left to right.

````python
def add(a, b):
    print(a + b)

add(10, 20)
add(5, 7)
````

````python
def show_student(name, age):
    print("Name:", name)
    print("Age:", age)

show_student("Ram", 20)
````

## 3. The `return` statement

The `return` statement sends a result from the function back to the part of the program that called it. This is useful when we want to store the answer in a variable or use it in another calculation.

`print()` displays something on the screen, while `return` gives a value back to the caller. They are not the same thing.

````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
print(result * 2)
````

### Returning more than one value

Python allows a function to return multiple values. They are returned together as a tuple, and we can unpack them into separate variables.

````python
def calculate(a, b):
    total = a + b
    difference = a - b
    product = a * b
    return total, difference, product

total, difference, product = calculate(10, 5)

print("Total:", total)
print("Difference:", difference)
print("Product:", product)
````

### `return` ends the function

As soon as Python reaches a `return` statement, the function stops running. The example checks for a negative number first. If the condition is true, the function returns immediately.

````python
def check_number(number):
    if number < 0:
        return "Negative number"

    return "Zero or positive"

print(check_number(-5))
print(check_number(10))
````

## 4. Default arguments

A **default argument** is a value assigned to a parameter in the function definition. If the caller leaves that argument out, Python uses the default value.

````python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Ram")
````

### Using more than one default argument

A function can have multiple parameters with default values. The caller can accept all the defaults or provide values to replace some or all of them.

One important rule: parameters without default values must come before parameters with default values. Otherwise, Python raises a `SyntaxError`.

````python
def student(name, age=18, course="Python"):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student("Ram")
student("Sita", 21)
student("Hari", 22, "Data Science")
````

The following is an example of **invalid syntax**. It is shown as a comment so that the notebook can still run from top to bottom:

```python
def example(a=10, b):
    return a + b
```

Python does not allow a required parameter (`b`) to come after a parameter with a default value (`a`).

## 5. Keyword arguments

With **keyword arguments**, we write the parameter name when calling a function. This makes a call easier to read and means the arguments do not have to be written in the same order as the parameters, as long as each keyword is used correctly.

````python
def student(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student(age=20, course="Python", name="Ram")
````

We can also mix positional and keyword arguments. Positional arguments must come before keyword arguments in a function call.

````python
def introduce(name, city, course="Python"):
    print(f"{name} is from {city} and is learning {course}.")

introduce("Sita", city="Kathmandu")
introduce("Hari", city="Pokhara", course="Data Science")
````

## 6. `*args` — variable-length positional arguments

Sometimes we do not know in advance how many positional arguments a function will receive. Adding `*args` to a parameter lets the function accept any number of extra positional arguments.

Inside the function, `args` is a **tuple**. The name `args` is a convention; the `*` is what gives the parameter this behaviour.

````python
def add_all(*args):
    print("Received values:", args)
    print("Sum:", sum(args))

add_all(10, 20)
add_all(1, 2, 3, 4, 5)
````

### Example: finding an average

The function below works with any number of scores. It also handles the case where no scores are supplied, so it does not try to divide by zero.

````python
def average(*numbers):
    if len(numbers) == 0:
        return 0

    return sum(numbers) / len(numbers)

print(average(10, 20, 30))
print(average(5, 15))
print(average())
````

A regular parameter can be placed before `*args`. This is useful when one value has a special meaning and the remaining values form a group.

````python
def show_scores(name, *scores):
    print("Student:", name)
    print("Scores:", scores)

show_scores("Ram", 80, 90, 85)
````

## 7. `**kwargs` — variable-length keyword arguments

`**kwargs` allows a function to receive any number of extra keyword arguments. Inside the function, these values are stored in a **dictionary**. The name `kwargs` is a convention; the double asterisk is the important part.

````python
def student_info(**kwargs):
    print(kwargs)

student_info(name="Ram", age=20, course="Python")
````

### Using `*args` and `**kwargs` together

A function can accept ordinary parameters, extra positional arguments, keyword-only parameters, and extra keyword arguments. A common order is:

```python
def function_name(required, *args, default=10, **kwargs):
    pass
```

Here, `default` is keyword-only because it comes after `*args`.

````python
def example(name, *args, level="beginner", **kwargs):
    print("Name:", name)
    print("Extra positional arguments:", args)
    print("Level:", level)
    print("Extra keyword arguments:", kwargs)

example("Ram", 10, 20, level="intermediate", age=20, city="Kathmandu")
````

## 8. Local and global scope

**Scope** means the part of a program where a variable can be accessed. Understanding scope helps prevent confusion when functions use variables with the same name.

### Local variables

A local variable is created inside a function. It is normally available only inside that function.

````python
def my_function():
    message = "Hello from inside the function"
    print(message)

my_function()
````

The next example would raise a `NameError` because `message` is local to `my_function()`. It is kept as a comment rather than executed, so the rest of the notebook can run without interruption.

```python
def my_function():
    message = "Hello"

my_function()
print(message)  # NameError: message is not defined
```

### Global variables

A global variable is created outside a function. A function can read a global variable, provided it has not been hidden by a local variable with the same name.

````python
name = "Ram"

def greet():
    print("Hello", name)

greet()
print(name)
````

### Changing a global variable with `global`

If we want to assign a new value to a global variable from inside a function, we use the `global` keyword. Without it, an assignment normally creates a local variable instead.

````python
count = 0

def increase():
    global count
    count += 1

increase()
increase()
print("Count:", count)
````

### The `nonlocal` keyword

`nonlocal` is used in a nested function to refer to a variable in the nearest enclosing function scope. It does not refer to a global variable.

````python
def outer():
    count = 0

    def inner():
        nonlocal count
        count += 1

    inner()
    inner()
    print("Count inside outer:", count)

outer()
````

## 9. Lambda functions

A **lambda function** is a small anonymous function. It is useful when we need a short function for a simple operation, often in one place.

**Syntax:**

```python
lambda parameters: expression
```

A lambda contains one expression. The result of that expression is returned automatically; we do not write a `return` statement inside it.

````python
square = lambda x: x * x

print(square(5))
print(square(8))
````

````python
add = lambda a, b: a + b

print(add(10, 20))
````

### Using a lambda as a sorting key

The `key` argument of `sort()` can receive a function that tells Python what value to use for sorting. Here, each student is represented by a tuple containing a name and a score. `student[1]` selects the score, which is the second item.

````python
students = [
    ("Ram", 80),
    ("Sita", 95),
    ("Hari", 70)
]

students.sort(key=lambda student: student[1])
print(students)
````

## 10. Related function concepts

The following ideas are also useful when learning functions. They build on the concepts covered above.

### Recursive functions

A recursive function calls itself to solve a smaller version of the same problem. It needs:

- A **base case**, which stops the recursion.
- A **recursive case**, which calls the function again with a simpler input.

This example calculates a factorial. For example, `5!` means `5 × 4 × 3 × 2 × 1`.

````python
def factorial(n):
    if n == 0 or n == 1:  # base case
        return 1

    return n * factorial(n - 1)  # recursive case

print(factorial(5))
````

### Docstrings

A **docstring** is a description written as the first statement inside a function. It explains what the function does and can describe its inputs and result. We can view it using the function's `__doc__` attribute.

````python
def calculate_area(length, width):
    """Return the area of a rectangle using its length and width."""
    return length * width

print(calculate_area(10, 5))
print(calculate_area.__doc__)
````

For a longer function, a docstring can describe the arguments and return value in more detail.

````python
def rectangle_area(length, width):
    """
    Calculate the area of a rectangle.

    Args:
        length: The length of the rectangle.
        width: The width of the rectangle.

    Returns:
        The area, calculated by multiplying length and width.
    """
    return length * width

print(rectangle_area.__doc__)
````

### Function annotations

Annotations let us document the expected types of parameters and the return value. They do not automatically check or convert the values at runtime; they are hints for readers and tools.

````python
def add_numbers(a: int, b: int) -> int:
    return a + b

print(add_numbers(10, 20))
print(add_numbers.__annotations__)
````

### Passing a function as an argument

In Python, functions are objects. This means we can pass a function to another function just like we pass a number or a string.

In this example, `square` is passed into `apply_function`, and `apply_function` calls it with the supplied value.

````python
def square(number):
    return number * number

def apply_function(func, value):
    return func(value)

print(apply_function(square, 5))
````
