# Python Tuples and Sets

Python has several built-in data structures for keeping groups of values together. Two of them are **tuples** and **sets**. They may look similar at first, but they are designed for different purposes.

- A **tuple** is an ordered collection. It allows duplicate values, and its contents cannot be changed after it is created.
- A **set** is a collection of unique values. It can be changed after creation, but it does not support indexing or slicing.

In this notebook, we will learn how to create and use tuples and sets, how tuple packing and unpacking work, and how to use common methods such as `count()`, `index()`, `add()`, `update()`, `discard()`, `remove()`, `union()`, and `intersection()`.

## Tuples

A tuple is an ordered collection of items. It can contain values of different data types and can include the same value more than once. Tuples are **immutable**, which means you cannot add, remove, or replace their elements after creation.

Tuples are usually written using parentheses `()` with items separated by commas.


```python
# An empty tuple
t1 = ()

# A tuple of integers
t2 = (10, 20, 30, 40)

# A tuple of strings
t3 = ("apple", "banana", "mango")

# A tuple with different data types
t4 = (10, "Python", 3.14, True)

print(t1)
print(t2)
print(t3)
print(t4)
```

### Creating a tuple with one item

There is one small detail to remember when creating a tuple with only one item: include a comma after the item. Without the comma, Python treats the value as an ordinary value inside parentheses, not as a tuple.



```python
single_item = ("Python",)
not_a_tuple = ("Python")

print(single_item)
print(type(single_item))

print(not_a_tuple)
print(type(not_a_tuple))
```

## Tuple packing

Tuple packing means putting multiple values together into a single tuple. You do not always need to write parentheses; commas are what make the values a tuple.

**Syntax:** `tuple_name = value1, value2, value3`



```python
# Tuple packing
student = "Ram", 20, "Computer Science"

print(student)
print(type(student))
```

## Tuple unpacking

Tuple unpacking is the reverse of packing. It means assigning the values in a tuple to separate variables in one statement.

The number of variables on the left should normally match the number of items in the tuple.

**Syntax:** `variable1, variable2, variable3 = tuple_name`



```python
student = ("Ram", 20, "Computer Science")

name, age, course = student

print(name)
print(age)
print(course)
```

### Unpacking with the `*` operator

Sometimes you may want to collect several values into one variable while unpacking. You can use `*` for this. The starred variable receives the remaining values as a list.



```python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

## Accessing tuple elements: indexing

Tuples are ordered, so each item has an index. Indexing starts at `0`. Negative indexes count from the end of the tuple, with `-1` referring to the last item.

**Syntax:** `tuple_name[index]`



```python
fruits = ("apple", "banana", "mango", "orange")

print(fruits[0])   # First item
print(fruits[2])   # Third item
print(fruits[-1])  # Last item
print(fruits[-2])  # Second-last item
```

## Tuple slicing

Slicing extracts a portion of a tuple. It returns a new tuple and does not change the original one.

**Syntax:** `tuple_name[start:stop:step]`

- `start` is the index where the slice begins (included).
- `stop` is the index where the slice ends (not included).
- `step` determines how many positions to move each time.



```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[::2])
print(numbers[::-1])
```

## `count()` method

The `count()` method returns how many times a specified value appears in a tuple.

**Syntax:** `tuple_name.count(value)`

If the value is not present, the method returns `0`.



```python
numbers = (10, 20, 10, 30, 10, 40)

print(numbers.count(10))
print(numbers.count(20))
print(numbers.count(50))
```

## `index()` method

The `index()` method returns the index of the **first occurrence** of a specified value in a tuple.

**Syntax:** `tuple_name.index(value)`

If the value does not exist, Python raises a `ValueError`. You can also provide optional `start` and `stop` positions to search within part of the tuple.

**Syntax with a search range:** `tuple_name.index(value, start, stop)`



```python
fruits = ("apple", "banana", "mango", "banana", "orange")

print(fruits.index("banana"))
print(fruits.index("mango"))

# Start searching from index 2
print(fruits.index("banana", 2))

# You can also set a stop position
print(fruits.index("banana", 2, 4))
```

## Other useful tuple operations

Here are a few built-in operations that work with tuples.

| Operation | Example | Result |
|---|---|---|
| Length | `len((10, 20, 30))` | `3` |
| Minimum | `min((10, 20, 30))` | `10` |
| Maximum | `max((10, 20, 30))` | `30` |
| Sum | `sum((10, 20, 30))` | `60` |
| Membership | `20 in (10, 20, 30)` | `True` |
| Concatenation | `(1, 2) + (3, 4)` | `(1, 2, 3, 4)` |
| Repetition | `(1, 2) * 2` | `(1, 2, 1, 2)` |



```python
numbers = (10, 20, 30)

print(len(numbers))
print(min(numbers))
print(max(numbers))
print(sum(numbers))
print(20 in numbers)
print((1, 2) + (3, 4))
print((1, 2) * 2)
```

## Sets

A set is a collection of unique elements. Sets are useful when you want to remove duplicate values or work with groups of values, such as finding what two groups have in common.

Sets are **mutable**, which means you can add and remove items after creating them. However, sets are unordered, so they do not support indexing or slicing. The order in which items appear when printed is not guaranteed.

Sets are usually written using curly braces `{}`. Each item must be hashable, so you can use values such as numbers, strings, and tuples containing only hashable values, but you cannot put a list or dictionary inside a set.


```python
# A set of numbers
numbers = {10, 20, 30, 40}

# Duplicate values are automatically removed
values = {10, 20, 10, 30, 20}

print(numbers)
print(values)
print(type(values))
```

### Creating an empty set

Use `set()` to create an empty set. Empty curly braces `{}` create an empty dictionary, not a set.



```python
empty_set = set()
empty_dictionary = {}

print(type(empty_set))
print(type(empty_dictionary))
```

## `add()` method

The `add()` method adds one element to a set.

**Syntax:** `set_name.add(element)`

If the element is already present, nothing extra is added because sets do not allow duplicates.



```python
fruits = {"apple", "banana"}

fruits.add("mango")
print(fruits)

# Adding an item that is already present does not create a duplicate
fruits.add("banana")
print(fruits)
```

## `update()` method

The `update()` method adds multiple elements from an iterable, such as a list, tuple, or another set.

**Syntax:** `set_name.update(iterable)`

You can pass more than one iterable to `update()`. All of their elements are added to the set, and duplicates are ignored.



```python
numbers = {10, 20}

numbers.update([30, 40, 50])
print(numbers)

# update() can accept more than one iterable
numbers.update((60, 70), {80, 90})
print(numbers)
```

## `remove()` method

The `remove()` method deletes a specified element from a set.

**Syntax:** `set_name.remove(element)`

If the element is not in the set, Python raises a `KeyError`. Use this method when you expect the item to be present and want Python to report an error if it is missing.



```python
numbers = {10, 20, 30, 40}

numbers.remove(20)
print(numbers)

# Uncomment the next line to see the KeyError:
# numbers.remove(100)
```

## `discard()` method

The `discard()` method also removes a specified element. The difference is that it does not raise an error if the element is not found; it simply leaves the set unchanged.

**Syntax:** `set_name.discard(element)`

This can be useful when you are not sure whether an item is in the set.



```python
numbers = {10, 20, 30, 40}

numbers.discard(20)
print(numbers)

# No error occurs if the value is not present
numbers.discard(100)
print(numbers)
```

## Set union: `union()`

The union of two sets contains every element that appears in either set. Duplicate elements appear only once in the result.

**Syntax:** `set1.union(set2)`

You can also use the `|` operator to perform a union.



```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.union(B))
print(A | B)
```

## Set intersection: `intersection()`

The intersection of two sets contains only the elements that are present in **both** sets.

**Syntax:** `set1.intersection(set2)`

You can also use the `&` operator to find the intersection.



```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.intersection(B))
print(A & B)
```

## A few more useful set operations

Along with the methods covered above, these operations are handy when comparing sets.

| Operation | Example | What it does |
|---|---|---|
| Difference | `A - B` | Items in `A` but not in `B` |
| Symmetric difference | `A ^ B` | Items in either set, but not in both |
| Subset | `A.issubset(B)` | Checks whether all items in `A` are in `B` |
| Superset | `A.issuperset(B)` | Checks whether `A` contains all items in `B` |
| Disjoint | `A.isdisjoint(B)` | Checks whether the sets have no common items |
| Length | `len(A)` | Counts the items in `A` |
| Membership | `item in A` | Checks whether an item is in the set |



```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A - B)
print(A ^ B)
print({1, 2}.issubset(A))
print(A.issuperset({1, 2}))
print(A.isdisjoint({7, 8}))
print(len(A))
print(3 in A)
```
