**Assignment Notes**

Strings are one of the most common data types in Python. We use them whenever we work with words, sentences, names, messages, or other text.

These notes cover how to create strings, access individual characters, work with parts of a string, format text, use escape characters, and apply some useful string methods. I have included small examples so it is easier to understand how each one works.

## 1. What is a String?

A **string** is a sequence of characters. It can contain letters, numbers, spaces, and symbols. In Python, strings are written inside quotation marks.

### Creating strings

```python
name = "Pratik"
message = 'Hello, Python!'
```

Python allows both single and double quotation marks. The opening and closing quotation marks should match.

````python
name = "Pratik"
message = 'Hello, Python!'

print(name)
print(message)
print(type(name))
````

**Output:**
```text
Pratik
Hello, Python!
<class 'str'>
```

A string can also be written using triple quotes. Triple quotes are useful for text that covers more than one line.

````python
paragraph = """Python is easy to learn.
It is useful for many tasks."""
print(paragraph)
````

**Output:**
```text
Python is easy to learn.
It is useful for many tasks.
```

## 2. String Indexing

**Indexing** means accessing one character from a string by using its position. Python starts counting from **0**, not 1.

For example, in the word `PYTHON`, `P` is at index 0, `Y` is at index 1, and so on. Python also supports negative indexing, which starts counting from the end. The last character has index `-1`.

### Index positions

| Character | P | Y | T | H | O | N |
|---|---:|---:|---:|---:|---:|---:|
| Positive index | 0 | 1 | 2 | 3 | 4 | 5 |
| Negative index | -6 | -5 | -4 | -3 | -2 | -1 |

````python
word = "PYTHON"

print(word[0])   # First character
print(word[3])   # Character at index 3
print(word[-1])  # Last character
print(word[-2])  # Second-last character
````

**Output:**
```text
P
H
N
O
```

Remember: trying to access an index that does not exist causes an `IndexError`.

## 3. String Slicing

**Slicing** is used to get a part of a string instead of just one character.

### Syntax

```python
string[start:stop:step]
```

- `start` is the index where slicing begins.
- `stop` is the position where slicing ends. The stop position is not included.
- `step` tells Python how many positions to move each time.

If we leave out `start`, `stop`, or `step`, Python uses default values.

````python
text = "Programming"

print(text[0:4])   # From index 0 up to (not including) 4
print(text[3:7])   # A section from the middle
print(text[:6])    # Beginning up to index 6
print(text[4:])    # Index 4 to the end
print(text[::2])   # Every second character
print(text[::-1])  # The string in reverse
````

**Output:**
```text
Prog
gram
Progra
ramming
Pormig
gnimmargorP
```

Slicing is useful when we need to extract a username, part of a sentence, or a section of other text.

## 4. f-Strings

An **f-string** is a simple way to place variables or expressions inside a string. Put the letter `f` before the opening quotation mark, then write variable names or expressions inside curly brackets `{}`.

### Syntax

```python
f"Text {variable}"
```

````python
name = "Pratik"
age = 17

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"Next year I will be {age + 1}.")
````

**Output:**
```text
My name is Pratik.
I am 17 years old.
Next year I will be 18.
```

f-strings make output easier to read because we do not have to join lots of separate strings using `+`.

## 5. Escape Characters

An **escape character** is a special character that starts with a backslash (`\`). It tells Python to treat the next character in a special way.

Escape characters are useful when we want to add a new line, a tab space, quotation marks, or a backslash inside a string.

### Common escape characters

| Escape character | Meaning | Example use |
|---|---|---|
| `\n` | New line | Moves text to the next line |
| `\t` | Horizontal tab | Adds a tab space |
| `\'` | Single quote | Adds `'` inside a string |
| `\"` | Double quote | Adds `"` inside a string |
| `\\` | Backslash | Displays one backslash |

### a. New line (`\n`)

The `\n` escape character moves the text after it to a new line.

````python
print("Hello\nWelcome to Python")
````

**Output:**
```text
Hello
Welcome to Python
```

### b. Tab (`\t`)

The `\t` escape character adds a horizontal tab space between text.

````python
print("Name:\tPratik")
print("Subject:\tPython")
````

**Output:**
```text
Name:   Pratik
Subject:        Python
```
*The exact width of a tab can depend on where the text is displayed.*

### c. Quotes inside strings

If a string is surrounded by double quotes, we can use `\"` to include a double quote inside it. We can use `\'` to include a single quote.

````python
print("She said, \"Python is interesting!\"")
print('It\'s a useful language.')
````

**Output:**
```text
She said, "Python is interesting!"
It's a useful language.
```

Another option is to use different quotation marks around the string. For example, single quotes can surround a string that contains double quotes.

````python
print('He said, "Hello!"')
````

**Output:**
```text
He said, "Hello!"
```

### d. Backslash (`\\`)

A backslash has a special purpose in Python strings. To display an actual backslash, write it twice as `\\`.

````python
print("C:\\Users\\Student\\Documents")
````

**Output:**
```text
C:\Users\Student\Documents
```

A raw string, written with `r` before the quotation mark, is another option for many paths because backslashes are treated literally:

```python
path = r"C:\Users\Student\Documents"
```

## 6. Useful String Methods

Python has built-in string methods that help us change, search, and separate text. A method is called using a dot after the string or variable name.

**Important:** String methods usually return a new string. They do not change the original string because strings in Python are immutable.

### a. `strip()`

The `strip()` method removes spaces (or specified characters) from the beginning and end of a string. It does not remove spaces from the middle.

**Syntax:**
```python
string.strip()
```

````python
message = "   Hello Python!   "
clean_message = message.strip()

print(clean_message)
print(message)  # Original string is unchanged
````

**Output:**
```text
Hello Python!
   Hello Python!   
```

### b. `split()`

The `split()` method divides a string into a list. By default, it splits the string wherever there is whitespace.

**Syntax:**
```python
string.split(separator)
```

The separator is optional. We can provide a comma, hyphen, or another character to decide where to split.

````python
sentence = "Python is fun to learn"
words = sentence.split()

print(words)

colors = "red,green,blue"
print(colors.split(","))
````

**Output:**
```text
['Python', 'is', 'fun', 'to', 'learn']
['red', 'green', 'blue']
```

### c. `join()`

The `join()` method joins items from an iterable, such as a list of strings, into one string. The string before `.join()` becomes the separator.

**Syntax:**
```python
"separator".join(iterable)
```

````python
words = ["Python", "is", "easy"]
result = " ".join(words)
print(result)

names = ["Asha", "Ravi", "Mina"]
print(", ".join(names))
````

**Output:**
```text
Python is easy
Asha, Ravi, Mina
```

The items being joined should be strings. If a list contains numbers, convert them to strings before using `join()`.

### d. `replace()`

The `replace()` method replaces one part of a string with another.

**Syntax:**
```python
string.replace(old, new, count)
```

- `old` is the text we want to replace.
- `new` is the replacement text.
- `count` is optional and limits how many replacements are made.

````python
sentence = "I like Java. Java is popular."
print(sentence.replace("Java", "Python"))
print(sentence.replace("Java", "Python", 1))
````

**Output:**
```text
I like Python. Python is popular.
I like Python. Java is popular.
```

### e. `find()`

The `find()` method searches for a character or piece of text and returns the index of its first occurrence. If the text is not found, it returns `-1`.

**Syntax:**
```python
string.find(value, start, end)
```

The `start` and `end` positions are optional.

````python
sentence = "Learn Python step by step"

print(sentence.find("Python"))
print(sentence.find("step"))
print(sentence.find("Java"))
````

**Output:**
```text
6
13
-1
```

`find()` is helpful when we need to check where a word appears in a string.

### f. `upper()`

The `upper()` method returns a copy of the string with all letters changed to uppercase.

**Syntax:**
```python
string.upper()
```

````python
text = "Hello Python"
print(text.upper())
````

**Output:**
```text
HELLO PYTHON
```

### g. `lower()`

The `lower()` method returns a copy of the string with all letters changed to lowercase.

**Syntax:**
```python
string.lower()
```

````python
text = "HELLO PYTHON"
print(text.lower())
````

**Output:**
```text
hello python
```

## 7. Quick Comparison of String Methods

| Method | What it does | Example |
|---|---|---|
| `strip()` | Removes spaces/selected characters from both ends | `" hi ".strip()` |
| `split()` | Splits a string into a list | `"a,b".split(",")` |
| `join()` | Joins strings using a separator | `"-".join(["a", "b"])` |
| `replace()` | Replaces part of a string | `"cat".replace("c", "b")` |
| `find()` | Returns the index of the first match, or `-1` | `"hello".find("e")` |
| `upper()` | Converts letters to uppercase | `"hello".upper()` |
| `lower()` | Converts letters to lowercase | `"HELLO".lower()` |

### A few useful points to remember

- String indexes start at `0`.
- Negative indexes count from the end.
- The stop index in slicing is not included.
- `\n` starts a new line, while `\t` adds a tab.
- Use `f"..."` to place variables inside strings.
- `split()` makes a list, while `join()` combines strings into one string.
- String methods return new values; they do not modify the original string.

## 8. Small Practice Example

This example combines a few methods from the notes. It removes extra spaces, changes the text to lowercase, splits it into words, and joins the words using hyphens.

````python
message = "   Python Makes Learning Fun   "

cleaned = message.strip()
small_letters = cleaned.lower()
words = small_letters.split()
final_text = "-".join(words)

print("Original:", message)
print("Cleaned:", cleaned)
print("Final:", final_text)
````

**Output:**
```text
Original:    Python Makes Learning Fun   
Cleaned: Python Makes Learning Fun
Final: python-makes-learning-fun
```
