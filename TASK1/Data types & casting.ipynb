### Data Type

A data type tells Python what kind of value a variable holds and what we can do with that value. For example, `10`, `10.5`, and `"10"` may look similar, but Python treats them as an integer, a float, and a string. Knowing the data type helps us use values correctly in a program.

#### Common Python data types

| **Data type** | **Description** | **Example** |
|---|---|---|
| `int` | Whole numbers | `10` |
| `float` | Decimal numbers | `3.14` |
| `complex` | Numbers with real and imaginary parts | `2 + 3j` |
| `str` | Text | `"Python"` |
| `bool` | True or false values | `True` |
| `list` | Ordered, changeable collection | `[1, 2, 3]` |
| `tuple` | Ordered, unchangeable collection | `(1, 2, 3)` |
| `range` | Sequence of numbers | `range(1, 6)` |
| `dict` | Key-value collection | `{"name": "Ram"}` |
| `set` | Collection of unique values | `{1, 2, 3}` |
| `NoneType` | Represents the absence of a value | `None` |

Here are some variables using different data types:

````python
a = 10
price = 99.50
complex_num = 2 + 3j
is_student = True
name = "Ram"
fruits = ["apple", "banana", "mango"]
coordinates = (10, 20, 30)
numbers = range(1, 6)
set_numbers = {1, 2, 2, 3, 4}
student = {"name": "Ram", "age": 20, "course": "Python"}
no_value = None

print(type(a))
print(type(price))
print(type(complex_num))
print(type(is_student))
print(type(name))
print(type(fruits))
print(type(coordinates))
print(type(numbers))
print(type(student))
print(type(set_numbers))
print(type(no_value))
````

**Output:**
```text
<class 'int'>
<class 'float'>
<class 'complex'>
<class 'bool'>
<class 'str'>
<class 'list'>
<class 'tuple'>
<class 'range'>
<class 'dict'>
<class 'set'>
<class 'NoneType'>
```

The output shows that Python can identify the type of each variable. Notice that the repeated value `2` appears only once in the set because sets keep unique values.

#### Mutable and immutable data types

Python data types can also be grouped by whether their values can be changed after they are created.

- **Mutable:** The object can be changed after it is created.
- **Immutable:** The object cannot be changed after it is created.

| **Mutable** | **Immutable** |
|---|---|
| `list` | `int` |
| `dict` | `float` |
| `set` | `bool` |
| `bytearray` | `str` |
| | `tuple` |
| | `bytes` |
| | `frozenset` |

For example, we can add an item to a list after creating it, but we cannot change one character inside a string directly.

````python
my_fruits = ["apple", "banana"]
my_fruits.append("mango")
print(my_fruits)

my_name = "Ram"
# Strings are immutable, so we create a new string when changing the value.
my_name = "Sam"
print(my_name)
````

**Output:**
```text
['apple', 'banana', 'mango']
Sam
```

The list was updated using `append()`. For the string, we assigned a new value to the variable instead of changing the original string.

### The `type()` Function

The `type()` function tells us the data type of a value or variable. It is useful when we are unsure what kind of data a variable contains.

**Syntax:**
```python
type(value)
```

**Example:**

````python
age = 20
name = "Ram"
height = 5.8
is_student = True
result = None

print(type(age))
print(type(name))
print(type(height))
print(type(is_student))
print(type(result))
````

**Output:**
```text
<class 'int'>
<class 'str'>
<class 'float'>
<class 'bool'>
<class 'NoneType'>
```

### Type Casting in Python

Type casting, also called type conversion, means changing a value from one data type to another. For example, a number entered through `input()` is received as text. If we want to calculate with it, we usually convert it into an integer or a float first.

Python supports built-in conversion functions such as `int()`, `float()`, `str()`, and `bool()`. It also has conversion functions for collections, such as `list()`, `tuple()`, and `set()`.

#### Types of type conversion

**1. Implicit type conversion**

Implicit conversion happens when Python automatically changes a value's type during an operation. For example, when an integer is added to a float, Python converts the integer to a float so the calculation can be completed.

**Example:**

````python
a = 10
b = 2.5
result = a + b

print(result)
print(type(result))
````

**Output:**
```text
12.5
<class 'float'>
```

**2. Explicit type conversion**

Explicit conversion happens when we use a conversion function ourselves to change a value's type. We choose the function depending on the type we need.

**Example:**

````python
age = "20"
age = int(age)
print(age + 5)
print(type(age))
````

**Output:**
```text
25
<class 'int'>
```

The value started as the string `"20"`. After using `int()`, it became the integer `20`, so Python could add 5 to it.

#### Common type-casting functions

| **Function** | **Conversion** | **Example** | **Result** |
|---|---|---|---|
| `int()` | To integer | `int("25")` | `25` |
| `float()` | To float | `float("3.5")` | `3.5` |
| `str()` | To string | `str(100)` | `"100"` |
| `bool()` | To Boolean | `bool(1)` | `True` |
| `list()` | To list | `list((1, 2))` | `[1, 2]` |
| `tuple()` | To tuple | `tuple([1, 2])` | `(1, 2)` |
| `set()` | To set | `set([1, 1, 2])` | `{1, 2}` |

The examples below focus on the four conversion functions required for this topic.

#### 1. Converting a string to an integer

We use `int()` when a string contains a whole-number value and we want to do arithmetic with it. The string must contain a valid integer.

````python
number = "100"
number = int(number)
print(number + 50)
````

**Output:**
```text
150
```

#### 2. Converting a value to a float

We use `float()` when we need a decimal number. It can convert an integer or a string containing a valid decimal number.

````python
price = "45.75"
price = float(price)
print(price)
print(type(price))
````

**Output:**
```text
45.75
<class 'float'>
```

#### 3. Converting a number to a string

We use `str()` to change a value into text. This is helpful when joining a number with a message.

````python
age = 20
message = "My age is " + str(age)
print(message)
````

**Output:**
```text
My age is 20
```

#### 4. Converting a float to an integer

When `int()` converts a float, it removes the decimal part. It does not round the number to the nearest whole number. If we want normal rounding, we can use `round()` instead.

````python
print(int(9.8))
print(int(-9.8))
print(round(9.8))
````

**Output:**
```text
9
-9
10
```

#### 5. Converting user input

The `input()` function returns text. If we want to add two numbers entered by a user, we should convert both inputs with `int()` or `float()` first.

````python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print(a + b)
````

**Sample output:**
```text
Enter first number: 10
Enter second number: 20
30
```

#### 6. Boolean type casting

The `bool()` function converts a value to either `True` or `False`. In general, zero, empty values, and `None` become `False`. Non-zero numbers and non-empty strings become `True`.

````python
print(bool(1))
print(bool(0))
print(bool(""))
print(bool("Python"))
print(bool(None))
````

**Output:**
```text
True
False
False
True
False
```

One thing that can be confusing at first is that `bool("False")` returns `True`. This is because `"False"` is still a non-empty string. Python checks whether the value is empty or zero, not what a word means.

#### Common false-like values in Python

These values are treated as `False` when used in a Boolean context:

- `0`
- `0.0`
- `""` (empty string)
- `[]` (empty list)
- `()` (empty tuple)
- `{}` (empty dictionary)
- `set()` (empty set)
- `None`

A string containing a space, such as `" "`, is not empty, so `bool(" ")` returns `True`.

````python
print(bool(""))
print(bool(" "))
print(bool([]))
print(bool({}))
print(bool(set()))
````

**Output:**
```text
False
True
False
False
False
```
