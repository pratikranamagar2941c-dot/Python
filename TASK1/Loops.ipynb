
# Loops in Python

Loops are one of the most useful parts of Python. They let us repeat a block of code without writing the same statements again and again.

For example, writing the same `print()` statement five times works, but it is not a good way to solve a problem:

```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```

A loop makes the same task shorter and easier to change:

```python
for i in range(5):
    print("Hello")
```

In this notebook, we will cover:

- `for` loops
- `while` loops
- `range()`
- `break`
- `continue`
- `pass`
- nested loops
- `enumerate()`
- `zip()`

The examples are kept simple at first and then become a little more practical.



## 1. `for` Loop

A `for` loop is used when we want to go through the items of something one by one.

A sequence can be a list, tuple, string, set, dictionary, or a `range()` of numbers.

### Basic syntax

```python
for variable in sequence:
    # code to repeat
```

The variable takes the value of one item at a time.


````python

fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
````

````python

for letter in "Python":
    print(letter)
````


### Example: Calculating a total

A loop is often useful when we need to process every value in a collection.

Here, `total` starts at zero and each number is added to it.


````python

numbers = [10, 20, 30, 40]
total = 0

for number in numbers:
    total = total + number

print("Total:", total)
````


## 2. `range()` Function

The `range()` function gives us a sequence of integers. It is especially useful when a loop needs to run a certain number of times.

There are three common forms:

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

One important rule to remember is that the **stop value is not included**.


````python

# range(stop)
for i in range(5):
    print(i)
````

````python

# range(start, stop)
for i in range(2, 6):
    print(i)
````

````python

# range(start, stop, step)
for i in range(2, 11, 2):
    print(i)
````

````python

# Counting backward with a negative step
for i in range(5, 0, -1):
    print(i)
````


Notice that `0` is not printed in the last example. The stop value is always excluded.

A quick way to remember `range()` is:

- `range(5)` → starts at `0`, stops before `5`
- `range(2, 6)` → starts at `2`, stops before `6`
- `range(2, 11, 2)` → starts at `2`, increases by `2`, stops before `11`



## 3. `while` Loop

A `while` loop keeps running **as long as its condition is `True`**.

### Basic syntax

```python
while condition:
    # code to repeat
```

Unlike a normal `for` loop, we usually need to update a variable ourselves. If the condition never becomes false, the loop can continue forever.


````python

i = 1

while i <= 5:
    print(i)
    i += 1
````


### `for` vs `while`

| `for` loop | `while` loop |
|---|---|
| Commonly used to iterate over a sequence or range | Commonly used when repetition depends on a condition |
| The next item is handled by the loop | We often update the control variable ourselves |
| Usually easier to keep finite | Easier to accidentally create an infinite loop |

For example, these two loops produce the same result:


````python

# Using for
for i in range(3):
    print(i)
````

````python

# Using while
i = 0

while i < 3:
    print(i)
    i += 1
````


### Infinite loops

An infinite loop happens when its condition never becomes `False`.

For example, this should **not** be run because `i` is never changed:

```python
# Don't run this
# i = 1
# while i <= 5:
#     print(i)
```

If you are writing a `while` loop, always check how and when its condition will eventually become false.



## 4. `break` Statement

`break` immediately stops the **nearest enclosing loop**.

It is useful when we have found what we were looking for and there is no reason to continue.


````python

for i in range(1, 10):
    if i == 5:
        break
    print(i)
````


### Example: Searching for an item

Once `30` is found, there is no need to keep checking the remaining numbers.


````python

numbers = [10, 20, 30, 40, 50]

for number in numbers:
    if number == 30:
        print("Number found")
        break
````


## 5. `continue` Statement

`continue` skips the rest of the **current iteration** and moves to the next iteration of the loop.

Think of it as: "Skip this one and keep going."


````python

for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````


### Example: Print only odd numbers

An even number has a remainder of `0` when divided by `2`, so we can skip it.


````python

for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
````


### `break` vs `continue`

- `break` → stop the loop completely.
- `continue` → skip the current iteration, but keep the loop running.



## 6. `pass` Statement

`pass` literally does nothing.

It is useful as a placeholder when Python requires a statement, but we are not ready to write the actual code yet.

This is different from `continue`: `continue` changes the flow of a loop, while `pass` simply performs no action.


````python

for i in range(5):
    pass

print("The loop ran, but pass did nothing.")
````

````python

# A common use while developing code
def future_function():
    pass
````


## 7. Nested Loops

A nested loop is simply a loop inside another loop.

For every iteration of the outer loop, the inner loop completes all of its iterations.

This is useful for working with grids, tables, patterns, and combinations of values.


````python

for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
````


### Example: Star pattern

The outer loop controls the number of rows. The inner loop controls how many stars appear in each row.


````python

for i in range(1, 5):
    for j in range(i):
        print("*", end="")
    print()
````


### Example: Multiplication tables

Nested loops can also be used to build a small multiplication table.


````python

for i in range(1, 3):
    for j in range(1, 4):
        print(f"{i} x {j} = {i * j}")
    print()
````


## 8. `enumerate()`

Sometimes we need both the **index** and the **value** while looping through a list.

`enumerate()` gives us both without needing to manually work with `range(len(...))`.

### Basic syntax

```python
for index, value in enumerate(iterable):
    # code
```

By default, counting starts from `0`.


````python

fruits = ["apple", "banana", "mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
````


### Without `enumerate()`

We could also write this:


````python

fruits = ["apple", "banana", "mango"]

for i in range(len(fruits)):
    print(i, fruits[i])
````


Both approaches work, but `enumerate()` is usually cleaner and communicates our intention more clearly.

### Starting the count at a different number

The optional `start` argument lets us choose where the numbering begins.


````python

students = ["Ram", "Sita", "Hari"]

for roll_no, student in enumerate(students, start=1):
    print(f"{roll_no}. {student}")
````


## 9. `zip()`

`zip()` lets us loop over two or more iterables at the same time.

It combines corresponding items into tuples.

### Basic syntax

```python
for item1, item2 in zip(iterable1, iterable2):
    # code
```


````python

names = ["Ram", "Sita", "Hari"]
ages = [20, 22, 21]

for name, age in zip(names, ages):
    print(name, age)
````


### Combining three lists

We can zip more than two iterables as well.


````python

names = ["Ram", "Sita", "Hari"]
ages = [20, 22, 21]
cities = ["Kathmandu", "Pokhara", "Lalitpur"]

for name, age, city in zip(names, ages, cities):
    print(f"{name}, {age}, {city}")
````


### What if the lists have different lengths?

By default, `zip()` stops as soon as the **shortest iterable** runs out of items.


````python

names = ["Ram", "Sita", "Hari"]
marks = [80, 90]

for name, mark in zip(names, marks):
    print(name, mark)
````


In the example above, `Hari` is not printed because there is no third mark.

This behavior is important to remember when the data being combined may not have the same length.



## 10. Combining the Ideas

The loop tools we learned are often used together.

For example, `enumerate()` is useful when we need a position, while `zip()` is useful when we need to process related lists together.


````python

students = ["Ram", "Sita", "Hari"]
marks = [80, 92, 75]

for position, (student, mark) in enumerate(zip(students, marks), start=1):
    print(f"{position}. {student} scored {mark}")
````
