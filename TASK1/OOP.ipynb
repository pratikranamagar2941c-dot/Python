# Object-Oriented Programming (OOP) in Python

**Assignment notes and practical examples**

Object-Oriented Programming, usually called **OOP**, is a way of writing programs by organising related data and actions into classes and objects. It helps keep code organised, reusable, and easier to maintain.

In this notebook, I have explained the main OOP concepts in a step-by-step way, with small examples that can be run and modified. The examples use familiar situations such as students, animals, and employees.

**Topics covered:** classes and objects, `self`, `__init__()`, methods, instance and class variables, static methods, inheritance, `super()`, method overriding, polymorphism, encapsulation, and Method Resolution Order (MRO).

## 1. Classes and objects

A **class** is like a blueprint or template. It describes the information and actions that objects created from it can have.

An **object** is an individual instance of a class. We can create many objects from the same class, and each object can store its own data.

### Basic syntax

```python
class ClassName:
    # attributes and methods
    pass

object_name = ClassName()
```

The `pass` statement is used when we want to define a class or function without adding its code yet.

````python
class Student:
    name = "Ram"
    age = 20

# Creating an object
s1 = Student()

print(s1.name)
print(s1.age)
````

We can create more than one object from the same class. Both objects have the same type, but they are separate objects.

````python
class Student:
    pass

s1 = Student()
s2 = Student()

print(type(s1))
print(type(s2))
print(s1 is s2)
````

The last result is `False` because `s1` and `s2` refer to two different objects, even though they were created from the same class.

## 2. The `__init__()` method

The `__init__()` method runs automatically when a new object is created. It is commonly called a constructor in beginner-level explanations. More precisely, it initialises the object's attributes after the object has been created.

It is useful when each object needs its own starting values.

### Syntax

```python
class ClassName:
    def __init__(self, parameter1, parameter2):
        self.attribute1 = parameter1
        self.attribute2 = parameter2
```

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Ram", 20)
s2 = Student("Sita", 22)

print(s1.name, s1.age)
print(s2.name, s2.age)
````

Here, `name` and `age` are parameters received when the object is created. The assignments using `self` store those values in the particular object.

If we do not define our own `__init__()` method, Python provides a default initialisation behaviour.

## 3. The `self` parameter

`self` refers to the current object (the instance) that is using a method. It allows an instance method to read or change that object's attributes and call its other instance methods.

The name `self` is a strong Python convention. Technically, another name could be used, but using `self` makes code easier for other Python programmers to understand.

When we call `s1.greet()`, Python automatically passes `s1` as the first argument to the method.

````python
class Student:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, my name is", self.name)

s1 = Student("Ram")
s2 = Student("Sita")

s1.greet()
s2.greet()
````

We can also see the instance being passed explicitly. This is usually not how we write calls in everyday code, but it helps explain what `self` does.

````python
class Student:
    def greet(self):
        print("Hello from the student object")

s1 = Student()

# These two calls are equivalent
s1.greet()
Student.greet(s1)
````

## 4. Methods in a class

A **method** is a function defined inside a class. Methods describe what an object or a class can do.

Python commonly uses three kinds of methods:

1. **Instance methods** use `self` and work with an individual object.
2. **Class methods** use `cls` and work with the class itself. They are marked with `@classmethod`.
3. **Static methods** do not automatically receive `self` or `cls`. They are marked with `@staticmethod`.

### 4.1 Instance methods

Instance methods can access or update the data belonging to a particular object.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

    def has_passed(self):
        return self.marks >= 40

s1 = Student("Ram", 85)
s1.display()
print("Passed:", s1.has_passed())
````

### 4.2 Class methods

A class method receives the class as its first parameter, conventionally named `cls`. It is created with the `@classmethod` decorator.

Class methods are useful when working with class-wide information or when providing an alternative way to create objects.

````python
class Student:
    school = "ABC College"

    @classmethod
    def show_school(cls):
        print("School:", cls.school)

Student.show_school()
````

### 4.3 Static methods

A static method is related to a class, but it does not need to access an individual object or the class itself. It is created with `@staticmethod`.

It is useful for a helper operation that logically belongs with the class.

````python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))

calculator = Calculator()
print(calculator.add(5, 7))
````

### Quick comparison of method types

| Method type | Decorator | First parameter | Common use |
|---|---|---|---|
| Instance method | None required | `self` | Work with an object's data |
| Class method | `@classmethod` | `cls` | Work with class-level data |
| Static method | `@staticmethod` | No automatic first parameter | Related helper operation |

## 5. Instance variables and class variables

Variables in a class can belong to individual objects or be shared at the class level.

### Instance variables

- Belong to a particular object.
- Are commonly created using `self` inside `__init__()` or another instance method.
- Different objects can have different values.

### Class variables

- Are defined directly inside the class, outside its methods.
- Are shared by instances unless an instance creates an attribute with the same name.
- Are useful for information that is common to the class.

````python
class Student:
    school = "ABC College"  # Class variable

    def __init__(self, name, age):
        self.name = name    # Instance variable
        self.age = age      # Instance variable

s1 = Student("Ram", 20)
s2 = Student("Sita", 22)

print(s1.name, s1.age)
print(s2.name, s2.age)
print(s1.school)
print(s2.school)
````

### Changing a class variable and shadowing it

Changing a class variable through the class changes the shared class-level value. Assigning the same attribute through one object creates or updates an instance attribute for that object; it does not change the class variable for every object.

````python
class Student:
    school = "ABC College"

s1 = Student()
s2 = Student()

Student.school = "National College"
print(s1.school)
print(s2.school)

s1.school = "Private Campus"
print(s1.school)          # Instance attribute
print(s2.school)          # Still reads the class attribute
print(Student.school)     # Class attribute
````

A common mistake is to use a mutable class variable, such as a list, when each object is supposed to have its own list. In that situation, create the list as an instance variable inside `__init__()`.

````python
class Classroom:
    def __init__(self, name):
        self.name = name
        self.students = []  # A separate list for each classroom

room_a = Classroom("A")
room_b = Classroom("B")

room_a.students.append("Ram")

print(room_a.students)
print(room_b.students)
````

## 6. Inheritance

**Inheritance** allows one class to reuse attributes and methods from another class. It is useful when several classes share common features.

- **Parent class (base class):** the class whose features are inherited.
- **Child class (derived class):** the class that inherits from the parent and can add or change features.

### Basic syntax

```python
class Parent:
    # shared attributes and methods
    pass

class Child(Parent):
    # additional or changed behaviour
    pass
```

````python
class Animal:
    def eat(self):
        print("The animal is eating")

class Dog(Animal):
    def bark(self):
        print("The dog is barking")

dog = Dog()
dog.eat()   # Inherited from Animal
dog.bark()  # Defined in Dog
````

### Common types of inheritance

**1. Single inheritance:** one child class inherits from one parent class.

````python
class Person:
    def introduce(self):
        print("I am a person")

class Student(Person):
    pass

Student().introduce()
````

**2. Multiple inheritance:** one child class inherits from more than one parent class.

````python
class Camera:
    def take_photo(self):
        print("Photo taken")

class Phone:
    def make_call(self):
        print("Calling someone")

class SmartPhone(Camera, Phone):
    pass

device = SmartPhone()
device.take_photo()
device.make_call()
````

**3. Multilevel inheritance:** a class inherits from a child class, forming a chain.

````python
class LivingThing:
    def breathe(self):
        print("Breathing")

class Animal(LivingThing):
    pass

class Dog(Animal):
    pass

Dog().breathe()
````

**4. Hierarchical inheritance:** multiple child classes inherit from the same parent class.

````python
class Animal:
    def eat(self):
        print("Eating food")

class Dog(Animal):
    pass

class Cat(Animal):
    pass

Dog().eat()
Cat().eat()
````

**5. Hybrid inheritance:** a combination of two or more inheritance patterns. Its behaviour can involve multiple inheritance and a shared parent class. Python uses MRO to decide the order in which classes are searched.

````python
class Person:
    def identify(self):
        print("Person")

class Student(Person):
    pass

class Employee(Person):
    pass

class TeachingAssistant(Student, Employee):
    pass

assistant = TeachingAssistant()
assistant.identify()
print([cls.__name__ for cls in TeachingAssistant.mro()])
````

## 7. The `super()` function

The `super()` function gives access to the next class in the Method Resolution Order (MRO). It is often used in a child class to call a parent class method, especially when initialising inherited attributes.

It is better to understand `super()` as following the MRO rather than simply meaning “call my parent”. This matters in multiple inheritance.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Animal:", self.name)

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def show(self):
        super().display()
        print("Breed:", self.breed)

dog = Dog("Tommy", "Labrador")
dog.show()
````

In this example, `super().__init__(name)` runs the next `__init__()` method in the MRO and sets up the `name` attribute. The child class then adds its own `breed` attribute.

## 8. Method overriding

**Method overriding** happens when a child class defines a method with the same name as a method in its parent class. When that method is called on a child object, Python uses the child's version.

Overriding lets a child class provide behaviour that is more suitable for that particular type of object.

````python
class Animal:
    def sound(self):
        print("The animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("The dog barks")

animal = Animal()
dog = Dog()

animal.sound()
dog.sound()
````

Sometimes we want to extend the parent's behaviour instead of replacing it completely. We can call the inherited implementation with `super()`.

````python
class Animal:
    def sound(self):
        print("An animal makes a sound")

class Dog(Animal):
    def sound(self):
        super().sound()
        print("A dog barks")

Dog().sound()
````

## 9. Polymorphism

**Polymorphism** means “many forms”. In OOP, it means that the same method name or operation can work with different object types, with each type providing its own behaviour.

### 9.1 Polymorphism through method overriding

Different classes can define a method with the same name. We can call that method in the same way while Python runs the implementation belonging to each object.

````python
class Dog:
    def sound(self):
        return "Bark"

class Cat:
    def sound(self):
        return "Meow"

class Cow:
    def sound(self):
        return "Moo"

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.sound())
````

### 9.2 Duck typing

Python also supports a style called **duck typing**. The idea is that an object can be used if it provides the behaviour we need. It does not always need to inherit from a particular class.

A common way to express this idea is: if an object behaves like the thing we need, we can use it for that task.

````python
class Car:
    def move(self):
        print("The car is driving")

class Person:
    def move(self):
        print("The person is walking")

def start_moving(item):
    item.move()

start_moving(Car())
start_moving(Person())
````

### 9.3 Operator overloading

Some operators behave differently depending on the types of their operands. For example, `+` adds numbers, joins strings, and combines lists. This is another familiar example of polymorphic behaviour.

````python
print(10 + 20)
print("Hello " + "World")
print([1, 2] + [3, 4])
````

Python classes can also define special methods such as `__add__()` to customise how operators work with objects. This is often called operator overloading.

````python
class BookCollection:
    def __init__(self, books):
        self.books = books

    def __add__(self, other):
        return BookCollection(self.books + other.books)

    def __str__(self):
        return ", ".join(self.books)

first = BookCollection(["Python Basics"])
second = BookCollection(["Learning OOP"])

combined = first + second
print(combined)
````

## 10. Encapsulation

**Encapsulation** means keeping related data and methods together in a class and providing a sensible way to access or update that data.

Python does not enforce private access in exactly the same way as some other languages. Instead, it uses naming conventions and features such as properties.

### 10.1 Public attributes

Public attributes can be accessed directly from outside the class. They have no leading underscore.

````python
class Student:
    def __init__(self, name):
        self.name = name

student = Student("Ram")
print(student.name)

student.name = "Sita"
print(student.name)
````

### 10.2 Protected-by-convention attributes

A single leading underscore, such as `_marks`, indicates that an attribute is intended for internal use. It is a convention, not a strict access restriction.

````python
class Student:
    def __init__(self, marks):
        self._marks = marks

student = Student(80)
print(student._marks)  # Accessible, but intended for internal use
````

### 10.3 Name-mangled attributes

A double leading underscore, such as `__marks`, triggers **name mangling**. Python changes the internal attribute name to include the class name. This helps avoid accidental name clashes in subclasses, but it is not absolute security or true privacy.

````python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    def display(self):
        print("Marks:", self.__marks)

student = Student(85)
student.display()

# The attribute is internally stored under a mangled name:
print(student._Student__marks)
````

### 10.4 Using `@property` to control access

A property lets us use method logic while accessing an attribute with normal attribute syntax. A setter can check a value before storing it.

This example makes sure student marks stay between 0 and 100.

````python
class Student:
    def __init__(self, marks):
        self.marks = marks

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if not 0 <= value <= 100:
            raise ValueError("Marks must be between 0 and 100")
        self._marks = value

student = Student(85)
print(student.marks)

student.marks = 90
print(student.marks)
````

The next cell demonstrates validation without interrupting the notebook with an unhandled error.

````python
try:
    student.marks = 120
except ValueError as error:
    print("Could not update marks:", error)

print("Current marks:", student.marks)
````

## 11. Method Resolution Order (MRO)

**Method Resolution Order**, usually shortened to **MRO**, is the order Python follows when searching for a method or attribute in a class hierarchy.

MRO is especially important with multiple inheritance, where a class can have more than one parent. Python uses the C3 linearisation algorithm to create a consistent search order.

We can inspect the MRO using:

```python
ClassName.mro()
```

or:

```python
ClassName.__mro__
```

````python
class A:
    def show(self):
        print("Method from A")

class B(A):
    pass

class C(B):
    pass

print([cls.__name__ for cls in C.mro()])
C().show()
````

### Understanding MRO with multiple inheritance

In the following example, `D` inherits from both `B` and `C`. Python searches according to the MRO, which respects the order of the parent classes while keeping the inheritance order consistent.

````python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")
        super().show()

class C(A):
    def show(self):
        print("C")
        super().show()

class D(B, C):
    def show(self):
        print("D")
        super().show()

print([cls.__name__ for cls in D.mro()])
D().show()
````

The output follows `D`, then `B`, then `C`, then `A`, and finally Python's built-in `object` class. The calls to `super()` move to the next class in that MRO.

## 12. Example combining the main OOP concepts

The following example brings several ideas together in a small employee system.

- `Employee` is the parent class.
- `Developer` and `Analyst` inherit from `Employee`.
- The `company` attribute is a class variable.
- `name` and `salary` are instance data.
- The salary is stored in a name-mangled attribute.
- `super()` is used to initialise the shared employee information.
- Both child classes override the `work()` method.
- The loop demonstrates polymorphism because the same method call produces different results.

````python
class Employee:
    company = "ABC Technologies"

    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    def work(self):
        print(self.name, "is working")

    def get_salary(self):
        return self.__salary

    @staticmethod
    def is_valid_salary(salary):
        return salary >= 0


class Developer(Employee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language

    def work(self):
        print(self.name, "is coding in", self.language)


class Analyst(Employee):
    def work(self):
        print(self.name, "is analysing data")


employees = [
    Developer("Ram", 50000, "Python"),
    Analyst("Sita", 45000)
]

for employee in employees:
    employee.work()
    print("Salary:", employee.get_salary())
    print("Company:", employee.company)
    print()
````

We can also use the static method to check a salary without creating an employee object.

````python
print(Employee.is_valid_salary(50000))
print(Employee.is_valid_salary(-100))
````
