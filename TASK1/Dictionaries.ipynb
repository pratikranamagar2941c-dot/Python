# Python Dictionaries

A dictionary is a built-in Python data structure used to store information as **key-value pairs**. It is useful when each value has a label that helps you identify it, such as a student's name and age, or a product's name and price.

Dictionaries are **mutable**, which means you can add, update, and remove items after creating one. In modern Python, dictionaries preserve insertion order. Each key must be unique, although values can be repeated.

In this notebook, we will learn how to create dictionaries, access and update their values, work with nested dictionaries, and use common methods such as `keys()`, `values()`, `items()`, `get()`, `update()`, and `pop()`.

## Creating a dictionary

Dictionaries are written using curly braces `{}`. Each item has a key and a value, separated by a colon `:`. Items are separated by commas.

**Syntax:** `dictionary_name = {key: value, key: value}`



```python
# An empty dictionary
student = {}

# A dictionary with string keys and different value types
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science",
    "is_enrolled": True
}

print(student)
print(type(student))
```

You can also create a dictionary using the `dict()` constructor. This is another way to build a dictionary, especially when the keys are valid keyword-style names.



```python
# Using dict() with keyword arguments
student = dict(name="Sita", age=21, course="Information Technology")

print(student)
```

## Understanding key-value pairs

Each dictionary item has two parts:

- **Key:** The name or label used to identify an item.
- **Value:** The information stored under that key.

For example, in `{"name": "Ram"}`, `"name"` is the key and `"Ram"` is the value.

Keys must be unique and must be hashable. Strings, numbers, and tuples containing only hashable values can be used as keys. Lists and dictionaries cannot be used as keys. If you write the same key more than once, the later value replaces the earlier one.



```python
details = {
    "name": "Ram",
    "age": 20,
    "name": "Ramesh"  # This replaces the earlier value for "name"
}

print(details)
```

## Accessing dictionary values

You can access a value by placing its key inside square brackets.

**Syntax:** `dictionary_name[key]`

If the key does not exist, using square brackets raises a `KeyError`.



```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science"
}

print(student["name"])
print(student["age"])
print(student["course"])
```

### Using `get()` to access values

The `get()` method also retrieves a value using its key. The difference is that it does not raise an error when the key is missing. By default, it returns `None`.

You can provide a second argument to choose what should be returned when the key is not found.

**Syntax:** `dictionary_name.get(key, default_value)`



```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science"
}

print(student.get("name"))
print(student.get("grade"))
print(student.get("grade", "Not available"))
```

Use square brackets when you expect a key to exist and want Python to report a missing key. Use `get()` when a key may be absent and you want a safe default instead.

## Adding and updating dictionary items

Dictionaries are mutable, so you can add new key-value pairs or change values that are already stored.

To add a new item, assign a value to a key that is not yet in the dictionary. To update an item, assign a new value to an existing key.

**Syntax:** `dictionary_name[key] = value`



```python
student = {
    "name": "Ram",
    "age": 20
}

# Add a new key-value pair
student["course"] = "Computer Science"

# Update an existing value
student["age"] = 21

print(student)
```

## `update()` method

The `update()` method adds items from another dictionary (or another suitable iterable of key-value pairs) to the current dictionary. If a key already exists, its value is updated. If it is new, the key-value pair is added.

**Syntax:** `dictionary_name.update(other_dictionary)`



```python
student = {
    "name": "Ram",
    "age": 20
}

student.update({"age": 21, "course": "Computer Science"})

print(student)
```

You can also pass keyword arguments to `update()` when the keys are valid keyword names.



```python
student = {"name": "Sita", "age": 19}

student.update(course="Information Technology", age=20)

print(student)
```

## Removing items with `pop()`

The `pop()` method removes the item with the specified key and returns its value.

**Syntax:** `dictionary_name.pop(key, default)`

- The `key` is required.
- The optional `default` is returned if the key does not exist.
- If the key is missing and no default is provided, Python raises a `KeyError`.



```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science"
}

removed_course = student.pop("course")

print("Removed value:", removed_course)
print("Updated dictionary:", student)
```

Here is what happens when the key is not present. Providing a default value prevents a `KeyError`.



```python
student = {"name": "Sita", "age": 20}

result = student.pop("grade", "No grade recorded")

print(result)
print(student)
```

## Dictionary methods: `keys()`, `values()`, and `items()`

These three methods let you view different parts of a dictionary.

### `keys()`

`keys()` returns a view containing all the keys in the dictionary.

**Syntax:** `dictionary_name.keys()`



```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science"
}

print(student.keys())
```

### `values()`

`values()` returns a view containing all the values in the dictionary.

**Syntax:** `dictionary_name.values()`



```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science"
}

print(student.values())
```

### `items()`

`items()` returns a view containing the dictionary's key-value pairs. Each pair is represented as a tuple.

**Syntax:** `dictionary_name.items()`



```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science"
}

print(student.items())
```

### Using dictionary views in loops

The `keys()`, `values()`, and `items()` methods are often used with `for` loops. This makes it easy to work with dictionary data one item at a time.



```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science"
}

# Loop through keys
for key in student.keys():
    print(key)

print("---")

# Loop through values
for value in student.values():
    print(value)

print("---")

# Loop through keys and values together
for key, value in student.items():
    print(key, ":", value)
```

These methods return **view objects**, not ordinary lists. A view reflects changes made to the dictionary. If you need a separate list, you can convert the view using `list()`.



```python
student = {"name": "Ram", "age": 20}

keys_list = list(student.keys())
values_list = list(student.values())
items_list = list(student.items())

print(keys_list)
print(values_list)
print(items_list)
```

## Nested dictionaries

A nested dictionary is a dictionary stored inside another dictionary. This is useful when you need to organise related information into groups, such as storing details for several students.



```python
students = {
    "student1": {
        "name": "Ram",
        "age": 20,
        "course": "Computer Science"
    },
    "student2": {
        "name": "Sita",
        "age": 21,
        "course": "Information Technology"
    }
}

print(students)
```

### Accessing values in a nested dictionary

To access a value inside a nested dictionary, use the outer key first, followed by the key inside the inner dictionary.

**Syntax:** `dictionary_name[outer_key][inner_key]`



```python
print(students["student1"]["name"])
print(students["student2"]["course"])
print(students["student1"]["age"])
```

### Updating nested dictionary values

You can update a value inside a nested dictionary by following the same key path and assigning a new value.



```python
students["student1"]["age"] = 22
students["student2"]["course"] = "Software Engineering"

print(students)
```

You can also add a new item to an inner dictionary.



```python
students["student1"]["city"] = "Kathmandu"

print(students["student1"])
```

## Checking whether a key exists

The `in` operator checks whether a key is present in a dictionary. It checks keys by default, not values.



```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Computer Science"
}

print("name" in student)
print("grade" in student)
print("Ram" in student)  # Checks keys, not values
print("Ram" in student.values())
```

## Other useful dictionary operations

Here are a few operations that are commonly used when working with dictionaries.

| Operation | Example | What it does |
|---|---|---|
| Length | `len({"a": 1, "b": 2})` | Returns the number of key-value pairs |
| Membership | `"a" in {"a": 1}` | Checks whether a key exists |
| Not in | `"c" not in {"a": 1}` | Checks whether a key is absent |
| Clear | `my_dict.clear()` | Removes all items |
| Copy | `my_dict.copy()` | Creates a shallow copy |
| Set default | `my_dict.setdefault("city", "Pokhara")` | Gets a value, adding the key with a default if missing |



```python
student = {"name": "Ram", "age": 20}

print(len(student))
print("name" in student)
print("grade" not in student)

copied_student = student.copy()
copied_student["age"] = 21

print("Original:", student)
print("Copy:", copied_student)

student.setdefault("city", "Pokhara")
print(student)
```
