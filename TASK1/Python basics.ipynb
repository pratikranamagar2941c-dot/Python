## 2. Indentation

Indentation means leaving spaces at the beginning of a line of code. Python uses indentation to show which statements belong together in a block. In most cases, four spaces are used.

For example, the `print()` statement below is indented because it belongs inside the `if` statement.

````python
if 10 > 5:
    print("10 is greater than 5")
````

**Output:**
```text
10 is greater than 5
```

If the line inside the `if` statement is not indented, Python will show an `IndentationError`. Indentation is important because it keeps code organized and tells Python how the program is structured.

Comments are notes that programmers add to explain their code. Python ignores comments when it runs a program. They are useful when we come back to our code later or when another person needs to understand it.

A single-line comment begins with the `#` symbol.

````python
# Display a welcome message
print("Welcome to Python")
````

**Output:**

For a note that covers several lines, we can put `#` at the beginning of each line. Triple-quoted strings are also used for documentation, although they are technically strings rather than ordinary comments.

````python
# This program introduces Python.
# It prints a short message on the screen.
print("I am learning Python.")
````

## 6. Variables

A variable is a name that refers to a value in a program. We use variables to store information so that we can use it again or change it later. In Python, we do not have to write the data type when creating a variable; Python works it out from the value we assign.

**Syntax:**
```python
variable_name = value
```

For example:

````python
student_name = "Patrick"
age = 18
average_marks = 85.5
is_student = True

print(student_name)
print(age)
print(average_marks)
print(is_student)
````

**Output:**
```text
Patrick
18
85.5
True
```

Python has several common data types:

| Data type | Example | What it stores |
|---|---|---|
| Integer (`int`) | `age = 18` | Whole numbers |
| Float (`float`) | `price = 25.5` | Decimal numbers |
| String (`str`) | `name = "Ram"` | Text |
| Boolean (`bool`) | `is_ready = True` | `True` or `False` |
| List (`list`) | `marks = [80, 90]` | A collection of values |

A variable's value can also be updated:

````python
score = 10
print("Old score:", score)

score = 20
print("New score:", score)
````

**Output:**
```text
Old score: 10
New score: 20
```

## 7. Variable Naming Rules

A variable name should follow Python's naming rules so that the program can understand it. Good names also make code easier for people to read.

The main rules are:
1. A name must start with a letter or an underscore (`_`).
2. It cannot start with a number.
3. It may contain letters, numbers, and underscores.
4. Spaces and special characters such as `@` and `$` are not allowed.
5. Python keywords, such as `class` and `if`, cannot be used as variable names.
6. Names are case-sensitive, so `name` and `Name` are different variables.

Examples:

````python
# Valid variable names
student_name = "Patrick"
age2 = 18
_total_marks = 450

print(student_name, age2, _total_marks)
````

Some examples of valid and invalid names are shown below:

| Valid names | Invalid names | Reason |
|---|---|---|
| `student_name` | `2student` | A name cannot start with a number. |
| `age` | `student age` | Spaces are not allowed. |
| `_marks` | `@marks` | Special characters are not allowed. |
| `total1` | `class` | `class` is a Python keyword. |

It is also a good habit to use clear names and lowercase letters. When a name has more than one word, we can separate the words with underscores. This style is called **snake_case**.
