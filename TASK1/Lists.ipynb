# Python Lists

A list is a built-in Python data structure used to store multiple values in a single variable. Lists are useful when you want to keep related items together and work with them as a group.

A list is **ordered**, allows **duplicate values**, and is **mutable**, which means you can change its contents after creating it. A list can also contain values of different data types.

In this notebook, we will look at how to create lists, access and update their elements, work with slices, and use common list methods.

## Creating lists

Lists are written using square brackets `[]`. Items are separated by commas.



```python
# An empty list
empty_list = []

# A list of integers
numbers = [10, 20, 30, 40]

# A list of strings
fruits = ["apple", "banana", "mango"]

# A list with different data types
mixed = [10, "Python", 3.14, True]

print(empty_list)
print(numbers)
print(fruits)
print(mixed)
```

You can also create a list from another iterable, such as a string or a tuple, by using `list()`.



```python
letters = list("hello")
values = list((1, 2, 3))

print(letters)
print(values)
```

## Accessing list elements: indexing

Each item in a list has an index. Python uses **zero-based indexing**, so the first item is at index `0`.

Negative indexing counts from the end of the list: `-1` refers to the last item, `-2` to the second-last item, and so on.

**Syntax:** `list_name[index]`



```python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[0])   # First item
print(fruits[2])   # Third item
print(fruits[-1])  # Last item
print(fruits[-2])  # Second-last item
```

If you try to access an index that does not exist, Python raises an `IndexError`.



```python
fruits = ["apple", "banana", "mango"]

# Uncomment the next line to see the IndexError:
# print(fruits[5])
```

## List slicing

Slicing is used to get a portion of a list. It creates a **new list** and does not change the original list.

**Syntax:** `list_name[start:stop:step]`

- `start` is the index where the slice begins (included).
- `stop` is the index where the slice ends (not included).
- `step` determines how many positions to move each time.

You can leave out `start`, `stop`, or `step` when you want to use their default values.



```python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])   # Items at indexes 1, 2 and 3
print(numbers[:3])    # From the beginning up to index 2
print(numbers[3:])    # From index 3 to the end
print(numbers[::2])   # Every second item
print(numbers[::-1])  # A reversed copy of the list
```

## Updating list elements

Lists are mutable, so you can replace an item by assigning a new value to its index.

**Syntax:** `list_name[index] = new_value`

You can also update several items at once by assigning a list to a slice.



```python
fruits = ["apple", "banana", "mango", "orange"]

fruits[1] = "grapes"
print(fruits)

# Replace the items at indexes 1 and 2
fruits[1:3] = ["kiwi", "pineapple"]
print(fruits)
```

## Adding items to a list

Python provides a few methods for adding items. The method you choose depends on whether you are adding one item or several items, and where you want them to appear.

### `append()`

`append()` adds **one item** to the end of a list.

**Syntax:** `list_name.append(item)`

If you append another list, that entire list becomes a single nested item.



```python
fruits = ["apple", "banana"]
fruits.append("mango")

print(fruits)

# A list appended to another list becomes one item
fruits.append(["orange", "grapes"])
print(fruits)
```

### `extend()`

`extend()` adds each item from an iterable to the end of the list. It is useful when you want to combine the contents of two lists.

**Syntax:** `list_name.extend(iterable)`

Unlike `append()`, it adds the elements individually rather than adding the iterable as one item.



```python
fruits = ["apple", "banana"]
more_fruits = ["mango", "orange"]

fruits.extend(more_fruits)

print(fruits)
```

### `insert()`

`insert()` adds an item at a chosen position. The items already in the list move to make room.

**Syntax:** `list_name.insert(index, item)`



```python
fruits = ["apple", "mango", "orange"]

fruits.insert(1, "banana")

print(fruits)
```

## Removing items from a list

There are different ways to remove items. Some methods remove an item by value, while others remove it by position.

### `remove()`

`remove()` deletes the **first occurrence** of the specified value.

**Syntax:** `list_name.remove(value)`

If the value is not in the list, Python raises a `ValueError`.



```python
fruits = ["apple", "banana", "mango", "banana"]

fruits.remove("banana")

print(fruits)  # Only the first "banana" is removed
```

### `pop()`

`pop()` removes and returns an item at a specified index. If you do not provide an index, it removes and returns the last item.

**Syntax:** `list_name.pop(index)`

The index is optional. Trying to pop from an empty list, or using an invalid index, raises an `IndexError`.



```python
numbers = [10, 20, 30, 40, 50]

removed_item = numbers.pop(1)
print("Removed item:", removed_item)
print("List now:", numbers)

last_item = numbers.pop()
print("Last item removed:", last_item)
print("List now:", numbers)
```

## Sorting a list: `sort()`

`sort()` arranges the items in the list. By default, it sorts in ascending order.

**Syntax:** `list_name.sort()`

You can use `reverse=True` to sort in descending order. The method changes the original list and returns `None`; it does not create a sorted copy.



```python
numbers = [40, 10, 50, 20, 30]

numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)
```

For strings, sorting is based on their character order (Unicode code points). When sorting mixed data types such as integers and strings together, Python 3 generally raises a `TypeError` because those values cannot be directly compared.



```python
names = ["Rita", "Aman", "Sita", "Bikash"]

names.sort()
print(names)

names.sort(reverse=True)
print(names)
```

## Reversing a list: `reverse()`

`reverse()` reverses the order of the items in the existing list. It does not sort the items.

**Syntax:** `list_name.reverse()`

Like `sort()`, `reverse()` changes the original list and returns `None`.



```python
numbers = [10, 40, 20, 50, 30]

numbers.reverse()
print(numbers)
```

## Counting items: `count()`

`count()` returns the number of times a particular value appears in a list.

**Syntax:** `list_name.count(value)`

If the value does not appear, the result is `0`.



```python
numbers = [10, 20, 10, 30, 10, 40]

print(numbers.count(10))
print(numbers.count(20))
print(numbers.count(50))
```

## Other useful list operations

Python also provides a few built-in operations that are commonly used with lists.

| Operation | Example | What it does |
|---|---|---|
| Length | `len([10, 20, 30])` | Returns the number of items |
| Minimum | `min([10, 20, 30])` | Returns the smallest item |
| Maximum | `max([10, 20, 30])` | Returns the largest item |
| Sum | `sum([10, 20, 30])` | Adds numeric items |
| Membership | `20 in [10, 20, 30]` | Checks whether an item exists |
| Not in | `50 not in [10, 20, 30]` | Checks whether an item is absent |
| Concatenation | `[1, 2] + [3, 4]` | Joins two lists into a new list |
| Repetition | `[1, 2] * 2` | Repeats the list |



```python
numbers = [10, 20, 30]

print(len(numbers))
print(min(numbers))
print(max(numbers))
print(sum(numbers))
print(20 in numbers)
print(50 not in numbers)
print([1, 2] + [3, 4])
print([1, 2] * 2)
```
