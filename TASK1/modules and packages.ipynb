# Python Modules and Packages

**Assignment notes with practical examples**

As Python programs grow, it becomes useful to keep related code in separate files. Python uses **modules** and **packages** to organise code and make it reusable. This notebook covers `import`, `from ... import`, aliases, custom modules, packages, `__init__.py`, the `math`, `random`, and `datetime` modules, and `pip`.

The examples use simple language and practical situations. Run the code cells from top to bottom.

## 1. What are modules and packages?

A **module** is a Python file (`.py`) containing reusable code, such as functions, variables, and classes. For example, `calculator.py` could contain functions for addition and subtraction.

A **package** is a directory that groups related Python modules. For example, a package named `school` could contain modules for greetings and student information.

In short:
- **Module:** one Python file.
- **Package:** a folder used to organise related Python modules.

## 2. Importing modules with `import`

The `import` statement makes code from another module available in the current program. We normally access its contents using the module name followed by a dot.

**Syntax:**
```python
import module_name
```

Python includes a standard-library module called `math`, so we can try a simple import right away.

````python
import math
print(math.sqrt(25))
print(math.pi)
````

### Example using a custom module

Imagine a file named `calculator.py` containing functions for addition and subtraction. The following cell creates that file so the example can run inside this notebook.

````python
from pathlib import Path
import importlib

calculator_code = '''
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
'''

Path("calculator.py").write_text(calculator_code, encoding="utf-8")
importlib.invalidate_caches()

import calculator
print(calculator.add(10, 5))
print(calculator.subtract(10, 5))
````

## 3. Using `from ... import`

Sometimes we only need selected functions or variables from a module. We can import them directly.

**Syntax:**
```python
from module_name import object_name
```

After importing an item this way, we can use its name directly without writing the module name each time.

````python
from calculator import add
print(add(10, 5))
````

We can import more than one item by separating their names with commas.

````python
from calculator import add, subtract
print(add(8, 2))
print(subtract(8, 2))
````

### Common import styles

| Import statement | How to use it |
|---|---|
| `import calculator` | `calculator.add(2, 3)` |
| `from calculator import add` | `add(2, 3)` |
| `from calculator import add, subtract` | `add(2, 3)` or `subtract(5, 2)` |
| `import calculator as c` | `c.add(2, 3)` |

## 4. Import aliases using `as`

An **alias** is another name given to an imported module or object. It can make a long name shorter and code easier to read.

**Syntax:**
```python
import module_name as short_name
```

````python
import math as m
print(m.sqrt(25))
print(m.pi)
````

Some commonly recognised aliases are:

```python
import pandas as pd
import numpy as np
```

These are conventions used with pandas and NumPy. An alias does not install a library; it only gives it a shorter name after it is available.

We can also give an imported function an alias.

````python
from math import factorial as fact
print(fact(5))
````

## 5. Creating your own custom modules

A **custom module** is a Python file that we create ourselves. It can contain functions, variables, or classes that we want to reuse in other programs.

This cell creates a module called `message_tools.py` in the notebook's working folder.

````python
message_code = '''
def welcome(name):
    return f"Welcome, {name}!"

def goodbye(name):
    return f"Goodbye, {name}!"

school_name = "Green Valley School"
'''

Path("message_tools.py").write_text(message_code, encoding="utf-8")
importlib.invalidate_caches()

import message_tools
print(message_tools.welcome("Ram"))
print(message_tools.goodbye("Ram"))
print(message_tools.school_name)
````

### Why create custom modules?

- We can reuse code in different programs.
- Large programs become easier to organise.
- Each module can focus on a particular task.
- Updating one module can be easier than searching through a very large program.

Python searches for modules in locations listed in `sys.path`. A custom module in the current working directory can usually be imported directly.

````python
import sys
print("Number of module search locations:", len(sys.path))
print("Current working directory:", Path.cwd())
````

## 6. What is a Python package?

A **package** is a directory that groups related Python modules. A package called `school` might contain modules for greetings and student information.

A simple structure looks like this:

```text
school/
    __init__.py
    greetings.py
    student.py
```

Let's create this structure from the notebook.

````python
package_dir = Path("school")
package_dir.mkdir(exist_ok=True)

(package_dir / "greetings.py").write_text(
    'def greet(name):\n'
    '    return f"Hello, {name}!"\n'
    '\n'
    'def goodbye(name):\n'
    '    return f"Goodbye, {name}!"\n',
    encoding="utf-8"
)

(package_dir / "student.py").write_text(
    'def get_student():\n'
    '    return "Student: Ram"\n',
    encoding="utf-8"
)

(package_dir / "__init__.py").write_text(
    '# Initialisation file for the school package\n',
    encoding="utf-8"
)

importlib.invalidate_caches()
print("Package created:", package_dir.resolve())
print("Files:", sorted(path.name for path in package_dir.iterdir()))
````

### Importing modules from a package

We can use the package name followed by the module name, or import selected functions from a module inside the package.

````python
from school.greetings import greet, goodbye
from school.student import get_student

print(greet("Ram"))
print(goodbye("Ram"))
print(get_student())
````

````python
from school import greetings
print(greetings.greet("Sita"))
````

## 7. The `__init__.py` file

The `__init__.py` file is associated with Python packages. It can initialise a package or make selected functions and classes available when someone imports the package.

For example, we can place imports inside `__init__.py`. Then users can import selected functions directly from the package.

The dot in `from .greetings import greet` means that `greetings` is a module in the current package. This is called a **relative import**.

````python
(package_dir / "__init__.py").write_text(
    'from .greetings import greet, goodbye\n'
    'from .student import get_student\n',
    encoding="utf-8"
)

importlib.invalidate_caches()
import school
importlib.reload(school)

print(school.greet("Ram"))
print(school.goodbye("Ram"))
print(school.get_student())
````

**Note:** In modern Python, a directory without `__init__.py` can sometimes work as a namespace package. However, `__init__.py` is still commonly used because it gives us a clear place to initialise a package and define its public imports.

## 8. The `math` module

The `math` module is part of Python's **standard library**. It provides mathematical functions and constants, so we do not need to install it separately with `pip`.

### Common functions in `math`

| Function | Purpose | Example result |
|---|---|---:|
| `math.sqrt(25)` | Square root | `5.0` |
| `math.ceil(4.2)` | Round upward | `5` |
| `math.floor(4.8)` | Round downward | `4` |
| `math.factorial(5)` | Factorial | `120` |
| `math.pow(2, 3)` | Raise a number to a power | `8.0` |
| `math.fabs(-10)` | Absolute value as a float | `10.0` |
| `math.gcd(12, 18)` | Greatest common divisor | `6` |
| `math.trunc(4.9)` | Remove fractional part | `4` |

````python
import math

number = 16
print("Square root:", math.sqrt(number))
print("Factorial:", math.factorial(5))
print("Ceiling:", math.ceil(4.1))
print("Floor:", math.floor(4.9))
print("Power:", math.pow(2, 3))
print("Absolute value:", math.fabs(-10))
print("GCD:", math.gcd(12, 18))
print("Truncated value:", math.trunc(4.9))
````

### Mathematical constants and trigonometric functions

The module provides constants such as `pi` and `e`. Trigonometric functions such as `sin()` and `cos()` use angles in **radians**, not degrees. We can convert degrees to radians with `math.radians()`.

````python
print("Pi:", math.pi)
print("Euler's number:", math.e)
print("Sine of 90 degrees:", math.sin(math.radians(90)))
print("Cosine of 0 degrees:", math.cos(math.radians(0)))
````

## 9. The `random` module

The `random` module generates **pseudorandom** values. It is useful for simple games, simulations, random selections, and practice programs. It is part of Python's standard library.

### Common functions in `random`

| Function | Purpose |
|---|---|
| `random.random()` | Float from `0.0` up to, but not including, `1.0` |
| `random.randint(a, b)` | Integer from `a` to `b`, including both endpoints |
| `random.randrange(a, b)` | Integer from `a` up to, but not including, `b` |
| `random.choice(sequence)` | Select one item from a sequence |
| `random.shuffle(list)` | Shuffle a list in place |
| `random.sample(sequence, k)` | Select `k` unique items |
| `random.uniform(a, b)` | Float between the two bounds |

The exact results can change each time we run these examples.

````python
import random

print("Random float:", random.random())
print("Random integer from 1 to 10:", random.randint(1, 10))
print("Random number from 1 to 9:", random.randrange(1, 10))
print("Random decimal:", random.uniform(1, 5))
````

### Selecting an item with `choice()`

The `choice()` function selects one item from a sequence, such as a list or tuple.

````python
fruits = ["apple", "banana", "mango", "orange"]
print("Selected fruit:", random.choice(fruits))
````

### Shuffling a list with `shuffle()`

The `shuffle()` function changes the order of items in a list. It modifies the original list and returns `None`.

````python
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(numbers)
````

### Selecting unique items with `sample()`

The `sample()` function returns a new list containing the requested number of unique selections. It does not change the original list.

````python
names = ["Ram", "Sita", "Hari", "Gita", "Mina"]
selected_names = random.sample(names, 3)

print("Original list:", names)
print("Selected names:", selected_names)
````

**A small note about randomness:** the `random` module is designed for general-purpose randomness, not security-sensitive work such as passwords or security tokens. Python provides the `secrets` module for those situations.

## 10. The `datetime` module

The `datetime` module is used to work with dates, times, and time differences. It is part of Python's standard library.

We can import the whole module or import individual classes such as `date`, `datetime`, and `timedelta`.

````python
import datetime
from datetime import date, datetime, timedelta
````

### Getting the current date and time

`datetime.now()` returns the current local date and time. The exact output depends on when and where the notebook is run.

````python
now = datetime.now()
print(now)
````

### Getting today's date

`date.today()` returns the current local date without the time.

````python
today = date.today()
print(today)
````

### Creating a specific date

We can create a date by passing the year, month, and day to `date()`. These values can also be accessed separately using `.year`, `.month`, and `.day`.

````python
birthday = date(2005, 5, 15)

print("Birthday:", birthday)
print("Year:", birthday.year)
print("Month:", birthday.month)
print("Day:", birthday.day)
````

### Formatting dates and times with `strftime()`

The `strftime()` method converts a date or time into a formatted string. The format codes tell Python which parts to display.

````python
now = datetime.now()

print("Date:", now.strftime("%d-%m-%Y"))
print("Time:", now.strftime("%H:%M:%S"))
print("Weekday:", now.strftime("%A"))
print("Month:", now.strftime("%B"))
````

### Common `strftime()` format codes

| Code | Meaning | Example |
|---|---|---|
| `%d` | Day of the month | `31` |
| `%m` | Month number | `10` |
| `%Y` | Four-digit year | `2026` |
| `%y` | Two-digit year | `26` |
| `%H` | Hour in 24-hour format | `20` |
| `%I` | Hour in 12-hour format | `08` |
| `%M` | Minute | `30` |
| `%S` | Second | `15` |
| `%A` | Full weekday name | `Saturday` |
| `%B` | Full month name | `October` |

The examples are illustrative; actual output depends on the date and time.

### Calculating date differences with `timedelta`

`timedelta` represents a duration or difference between dates and times. We can use it to add or subtract days, weeks, hours, and other time units.

````python
today = date.today()

tomorrow = today + timedelta(days=1)
last_week = today - timedelta(days=7)
thirty_days_later = today + timedelta(days=30)

print("Today:", today)
print("Tomorrow:", tomorrow)
print("Seven days ago:", last_week)
print("Thirty days from now:", thirty_days_later)
````

We can also find the number of days between two dates by subtracting one date from another.

````python
start_date = date(2026, 1, 1)
end_date = date(2026, 1, 15)

difference = end_date - start_date
print("Number of days between the dates:", difference.days)
````

## 11. What is `pip`?

`pip` is Python's package installer. It is used to install, update, and remove external Python packages.

Python already includes a standard library with modules such as `math`, `random`, and `datetime`. These do not need to be installed using `pip`.

External libraries, such as `pandas`, `numpy`, and `requests`, are commonly installed with `pip` when they are not already available in the environment.

### Common `pip` commands

| Command | Purpose |
|---|---|
| `python -m pip --version` | Display the pip version |
| `python -m pip install pandas` | Install a package |
| `python -m pip uninstall pandas` | Uninstall a package |
| `python -m pip show pandas` | Show information about a package |
| `python -m pip list` | List installed packages |
| `python -m pip freeze` | List installed packages and versions |
| `python -m pip install --upgrade pandas` | Upgrade a package |

Using `python -m pip` helps ensure pip runs for the Python interpreter selected by the `python` command.
